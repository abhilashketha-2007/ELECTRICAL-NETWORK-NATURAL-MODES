```python

# ============================================================
# CELL 1: IMPORTS
# ============================================================

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

from IPython.display import display, Markdown

# Numerical linear algebra
from numpy.linalg import eig, inv

# Pandas is useful for clean result tables
import pandas as pd

# Improve numerical display
np.set_printoptions(precision=6, suppress=True)

print("Libraries imported successfully.")
```

```python

# ============================================================
# CELL 2: USER CONFIGURATION
# ============================================================

# ------------------------------------------------------------
# CAPACITORS
# ------------------------------------------------------------
# Units: Farads
C1 = 100e-9       # 100 nF
C2 = 100e-9       # 100 nF
C3 = 100e-9       # 100 nF


# ------------------------------------------------------------
# LOCAL INDUCTORS
# ------------------------------------------------------------
# Units: Henry
L1 = 10e-6        # 10 uH
L2 = 10e-6        # 10 uH
L3 = 10e-6        # 10 uH


# ------------------------------------------------------------
# SHARED INDUCTORS
# ------------------------------------------------------------
# M12 is shared between loops 1 and 2
# M23 is shared between loops 2 and 3
#
# Units: Henry
M12 = 2e-6        # 2 uH
M23 = 2e-6        # 2 uH


# ------------------------------------------------------------
# OPTIONAL: EXCITATION USED FOR RESONANCE PLOT
# ------------------------------------------------------------
# The natural modes themselves do NOT require an external source.
#
# For the forced-response demonstration, we apply a voltage
# excitation to loop 1.
#
# Units: Volts
V_drive = np.array([
    1.0,
    0.0,
    0.0
])


# ------------------------------------------------------------
# FREQUENCY SWEEP SETTINGS
# ------------------------------------------------------------
# These determine the graph showing resonance.
f_min = 1e3       # Hz
f_max = 300e3     # Hz
num_points = 5000


# ------------------------------------------------------------
# BASIC INPUT VALIDATION
# ------------------------------------------------------------

all_capacitors = [C1, C2, C3]
all_inductors = [L1, L2, L3, M12, M23]

if any(C <= 0 for C in all_capacitors):
    raise ValueError("All capacitances must be positive.")

if any(L <= 0 for L in all_inductors):
    raise ValueError("All inductances must be positive.")

if f_min <= 0 or f_max <= f_min:
    raise ValueError("Frequency limits are invalid.")

print("Input configuration loaded successfully.")
```

```python

# ============================================================
# CELL 3: CONSTRUCT L AND C MATRICES
# ============================================================

# Capacitance matrix
#
# Since the capacitors belong to individual loops:
#
# C = diag(C1, C2, C3)

C_matrix = np.diag([
    C1,
    C2,
    C3
])


# ------------------------------------------------------------
# INDUCTANCE MATRIX
# ------------------------------------------------------------
#
# For two physically shared inductors:
#
#          M12                 M23
#    Loop 1 <----> Loop 2 <----> Loop 3
#
# The mesh inductance matrix becomes
#
# [ L1+M12       -M12          0       ]
# [ -M12     L2+M12+M23      -M23      ]
# [   0          -M23       L3+M23     ]
#
# The negative off-diagonal terms arise because the two mesh
# currents pass through a shared branch in opposite reference
# directions.

L_matrix = np.array([
    [L1 + M12,       -M12,                0],
    [-M12,           L2 + M12 + M23,     -M23],
    [0,              -M23,                L3 + M23]
])


C_inverse = inv(C_matrix)


print("Capacitance matrix C:")
print(C_matrix)

print("\nInductance matrix L:")
print(L_matrix)

print("\nInverse capacitance matrix C^(-1):")
print(C_inverse)
```

```python
# ============================================================
# CELL 4: SYMBOLIC MATHEMATICAL MODEL
# ============================================================

# Define symbols
l1, l2, l3 = sp.symbols('L_1 L_2 L_3', positive=True)
m12, m23 = sp.symbols('M_12 M_23', positive=True)
c1, c2, c3 = sp.symbols('C_1 C_2 C_3', positive=True)

# Symbolic matrices

L_sym = sp.Matrix([
    [l1 + m12,       -m12,             0],
    [-m12,           l2 + m12 + m23,  -m23],
    [0,              -m23,             l3 + m23]
])

C_inv_sym = sp.diag(
    1/c1,
    1/c2,
    1/c3
)

print("Symbolic inductance matrix L:")
display(L_sym)

print("\nSymbolic inverse capacitance matrix C^(-1):")
display(C_inv_sym)

print("\nNatural-mode equation:")
display(
    sp.Eq(
        C_inv_sym * sp.Matrix(sp.symbols('v1 v2 v3')),
        sp.Symbol('omega')**2 * L_sym * sp.Matrix(sp.symbols('v1 v2 v3'))
    )
)
```

```python
# ============================================================
# CELL 5: EIGENVALUE / NATURAL FREQUENCY ENGINE
# ============================================================

def calculate_natural_modes(L, C):
    """
    Calculate natural frequencies and mode shapes.

    Mathematical problem:
        C^(-1) v = omega^2 L v

    We transform it to the symmetric ordinary eigenvalue problem:
        L^(-1/2) C^(-1) L^(-1/2) u = omega^2 u

    Then:
        v = L^(-1/2) u
    """

    # Eigen-decomposition of L
    L_eigenvalues, L_eigenvectors = np.linalg.eigh(L)

    # Check positive definiteness
    if np.any(L_eigenvalues <= 0):
        raise ValueError(
            "The inductance matrix L must be positive definite."
        )

    # Construct L^(-1/2)
    L_inv_sqrt = (
        L_eigenvectors
        @ np.diag(1 / np.sqrt(L_eigenvalues))
        @ L_eigenvectors.T
    )

    # Symmetric standard eigenvalue matrix
    A = L_inv_sqrt @ inv(C) @ L_inv_sqrt

    # Solve eigenvalue problem
    omega_squared, U = np.linalg.eigh(A)

    # Numerical round-off can sometimes produce tiny negative values
    # such as -1e-12 for theoretically zero values.
    omega_squared = np.maximum(omega_squared, 0)

    # Natural angular frequencies
    omega = np.sqrt(omega_squared)

    # Natural frequencies in Hz
    frequencies = omega / (2 * np.pi)

    # Recover original mode vectors
    modes = L_inv_sqrt @ U

    # Normalize each mode so its largest component has magnitude 1
    for k in range(modes.shape[1]):
        max_component = np.max(np.abs(modes[:, k]))

        if max_component > 0:
            modes[:, k] /= max_component

        # Make the largest component positive for consistent display
        index = np.argmax(np.abs(modes[:, k]))

        if modes[index, k] < 0:
            modes[:, k] *= -1

    return omega_squared, omega, frequencies, modes


# Run the eigenvalue calculation
omega_squared, omega, frequencies, modes = calculate_natural_modes(
    L_matrix,
    C_matrix
)

print("Eigenvalue calculation completed.")
```

```python
# ============================================================
# CELL 6: RESULTS TABLE
# ============================================================

results = pd.DataFrame({
    "Mode": [1, 2, 3],
    "Eigenvalue ω² (rad²/s²)": omega_squared,
    "Angular frequency ω (rad/s)": omega,
    "Natural frequency f (Hz)": frequencies,
    "Natural frequency f (kHz)": frequencies / 1e3
})

print("NATURAL FREQUENCIES")
display(results)


# ------------------------------------------------------------
# MODE SHAPES
# ------------------------------------------------------------

mode_table = pd.DataFrame(
    modes,
    index=["Loop 1", "Loop 2", "Loop 3"],
    columns=["Mode 1", "Mode 2", "Mode 3"]
)

print("\nNORMALIZED EIGENVECTORS / MODE SHAPES")
display(mode_table)
```

```python
# ============================================================
# CELL 7: EIGENVALUE VERIFICATION
# ============================================================

print("VERIFYING C^(-1) v = ω² L v\n")

for k in range(3):

    v = modes[:, k]
    lam = omega_squared[k]

    left_side = C_inverse @ v
    right_side = lam * L_matrix @ v

    error = np.linalg.norm(left_side - right_side)

    print(f"Mode {k + 1}")
    print("-" * 40)
    print("Left side  C^(-1)v:")
    print(left_side)

    print("\nRight side ω²Lv:")
    print(right_side)

    print(f"\nNumerical error = {error:.3e}\n")
```

```python
# ============================================================
# CELL 8: HUMAN-READABLE MATHEMATICAL MODEL
# ============================================================

print("THREE-LOOP NATURAL-MODE EQUATION")
print("=" * 60)

print("""
L q¨ + C^(-1) q = 0

where

q = [q1, q2, q3]^T
""")

print("Numerical L matrix:")
print(L_matrix)

print("\nNumerical C^(-1) matrix:")
print(C_inverse)

print("\nTherefore:")

print("\n     L q¨ + C^(-1) q = 0")

print("\nAssume:")
print("     q(t) = v exp(jωt)")

print("\nThen:")
print("     [C^(-1) - ω²L]v = 0")

print("\nFor a non-zero mode shape v:")
print("     det(C^(-1) - ω²L) = 0")

print("\nTherefore the roots give the natural frequencies.")
```

```python
# ============================================================
# CELL 11: FORCED RESPONSE / RESONANCE CALCULATION
# ============================================================

def calculate_frequency_response(
    frequencies_hz,
    L,
    C,
    drive_vector,
    damping=0.0
):
    """
    Calculate forced frequency response.

    Undamped equation:
        (C^(-1) - omega^2 L) q = V

    A small damping parameter can be introduced numerically
    to prevent an infinite response exactly at resonance.

    damping is dimensionless and is used here as a small
    regularizing term.
    """

    C_inv = inv(C)

    response = []

    for f in frequencies_hz:

        omega_f = 2 * np.pi * f

        # Dynamic stiffness matrix
        D = C_inv - omega_f**2 * L

        # Small numerical regularization for plotting
        if damping > 0:
            D = D + 1j * damping * np.eye(3)

        try:
            q = np.linalg.solve(D, drive_vector)

        except np.linalg.LinAlgError:
            # Exactly singular in the ideal undamped case
            q = np.full(3, np.nan, dtype=complex)

        response.append(q)

    return np.array(response)


# Frequency range
frequency_sweep = np.linspace(
    f_min,
    f_max,
    num_points
)


# Use a very small damping value for a realistic finite plot
response = calculate_frequency_response(
    frequency_sweep,
    L_matrix,
    C_matrix,
    V_drive,
    damping=1e-4
)


# Magnitude of each loop response
response_magnitude = np.abs(response)
```

```python
# ============================================================
# CELL 19: DYNAMIC PROJECT FUNCTION
# ============================================================

def solve_three_loop_network(
    C_values,
    L_values,
    shared_values
):
    """
    Solve a 3-loop LC network.

    Parameters
    ----------
    C_values : list/array
        [C1, C2, C3] in Farads

    L_values : list/array
        [L1, L2, L3] in Henry

    shared_values : list/array
        [M12, M23] in Henry
    """

    C1_, C2_, C3_ = C_values
    L1_, L2_, L3_ = L_values
    M12_, M23_ = shared_values

    C = np.diag([
        C1_,
        C2_,
        C3_
    ])

    L = np.array([
        [L1_ + M12_,       -M12_,                  0],
        [-M12_,            L2_ + M12_ + M23_,    -M23_],
        [0,                -M23_,                  L3_ + M23_]
    ])

    eigenvalues, omega, freq, modes = calculate_natural_modes(
        L,
        C
    )

    return {
        "C_matrix": C,
        "L_matrix": L,
        "eigenvalues": eigenvalues,
        "omega": omega,
        "frequency": freq,
        "modes": modes
    }


# ------------------------------------------------------------
# Example dynamic call
# ------------------------------------------------------------

new_result = solve_three_loop_network(
    C_values=[100e-9, 100e-9, 100e-9],
    L_values=[10e-6, 10e-6, 10e-6],
    shared_values=[2e-6, 2e-6]
)

print("Dynamic calculation complete.")

print("\nNatural frequencies:")
for i, f in enumerate(new_result["frequency"]):
    print(f"Mode {i+1}: {f/1000:.3f} kHz")
```

```python
# ============================================================
# CELL 22: ENERGY DISTRIBUTION TABLE
# ============================================================

# For a natural mode, assume the maximum capacitor charge
# amplitude is represented by the eigenvector.
#
# Capacitor energy:
#       E_C = (1/2) q^T C^(-1) q
#
# Inductor energy:
#       E_L = (1/2) i^T L i
#
# For a natural oscillation, the maximum electric and
# magnetic energies are equal.

energy_data = []

for k in range(3):

    v = modes[:, k]

    # Normalize mode vector
    v = v / np.max(np.abs(v))

    # Electric / capacitor energy
    electric_energy = 0.5 * v.T @ C_inverse @ v

    # At natural resonance, maximum magnetic energy is equal
    # to maximum electric energy.
    magnetic_energy = electric_energy

    total_energy = electric_energy + magnetic_energy

    # Energy percentage in each capacitor
    capacitor_energy = (
        0.5 * v**2 / np.array([C1, C2, C3])
    )

    capacitor_percent = (
        capacitor_energy / np.sum(capacitor_energy) * 100
    )

    energy_data.append({
        "Mode": k + 1,
        "Frequency (kHz)": frequencies[k] / 1000,
        "Electric Energy": electric_energy,
        "Magnetic Energy": magnetic_energy,
        "Total Energy": total_energy,
        "Loop 1 Energy (%)": capacitor_percent[0],
        "Loop 2 Energy (%)": capacitor_percent[1],
        "Loop 3 Energy (%)": capacitor_percent[2]
    })


energy_table = pd.DataFrame(energy_data)

display(energy_table)
```

```python
# ============================================================
# CELL 29: COUPLED vs UNCOUPLED COMPARISON
# ============================================================

# ------------------------------------------------------------
# UNCOUPLED SYSTEM
# ------------------------------------------------------------

L_uncoupled = np.diag([
    L1,
    L2,
    L3
])

uncoupled_lambda, uncoupled_omega, uncoupled_freq, uncoupled_modes = \
    calculate_natural_modes(
        L_uncoupled,
        C_matrix
    )


# ------------------------------------------------------------
# COUPLED SYSTEM
# ------------------------------------------------------------

coupled_lambda, coupled_omega, coupled_freq, coupled_modes = \
    calculate_natural_modes(
        L_matrix,
        C_matrix
    )


# ------------------------------------------------------------
# COMPARISON TABLE
# ------------------------------------------------------------

comparison_table = pd.DataFrame({

    "Mode": [1, 2, 3],

    "Uncoupled Frequency (kHz)":
        uncoupled_freq / 1000,

    "Coupled Frequency (kHz)":
        coupled_freq / 1000,

    "Frequency Shift (kHz)":
        (coupled_freq - uncoupled_freq) / 1000,

    "Frequency Shift (%)":
        ((coupled_freq - uncoupled_freq)
         / uncoupled_freq) * 100
})

display(comparison_table)
```
